```python
# Binomial distribution
from scipy.stats import binom

n=5
k=3
p=0.5

prob = binom.pmf(k, n, p)
print(prob)
```

```python
# Uniform distribution
import numpy as np
import matplotlib.pyplot as plt

# values from 1 to 10
values = np.random.uniform(0, 10, 10_000_000)

plt.hist(values, bins=100, density=True, alpha=0.3)

plt.title("Continuous Uniform Distribution between 0 and 10")
plt.xlabel("x")
plt.ylabel("Probability distribution")
plt.grid(True)
plt.show()
```

```python
# Normal Distribution
from scipy.stats import norm

# parameters
mu = 70    # mean
sigma = 10 # standard deviation

# x-axis values
x = np.linspace(30, 110, 1000)

# normal pdf
y = norm.pdf(x, mu, sigma)

# bell curve
plt.plot(x, y, color="black", linewidth=2, label="Normal Distribution")

# shade 68% region (μ +- σ)
x1 = np.linspace(mu - sigma, mu + sigma, 1000)
plt.fill_between(x1, norm.pdf(x1, mu, sigma), color="green", alpha=0.5, label="68% region (1σ)")

# shade 95% region (μ +- 2σ)
x2 = np.linspace(mu - 2*sigma, mu + 2*sigma, 1000)
plt.fill_between(x2, norm.pdf(x2, mu, sigma), color="blue", alpha=0.2, label="95% region (2σ)")

# shade 95% region (μ +- 3σ)
x3 = np.linspace(mu - 3*sigma, mu + 3*sigma, 1000)
plt.fill_between(x3, norm.pdf(x3, mu, sigma), color="pink", alpha=0.3, label="99.7% region (3σ)")


plt.title("Normal Distribution of Student Scores")
plt.xlabel("Score")
plt.ylabel("Probability Density")
plt.legend()
plt.show()
```
