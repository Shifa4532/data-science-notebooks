```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 100, 10_00_000)
y = np.sqrt(x)

plt.plot(x, y, label="y = √x")
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = √x")
plt.legend()
plt.grid(True)

plt.show()
```

```python
x = np.linspace(-2*np.pi, 2*np.pi, 100)
y = np.sin(x)

plt.plot(x, y, label="sin(x)", color="red")

plt.axvline(0, linewidth=1.5, color="black", linestyle="--", label="x=0")
plt.axhline(0, linewidth=1.5, color="black", linestyle="--", label="y=0")

plt.xlabel("x")
plt.ylabel("y")
plt.title("sin(x)")
plt.legend()
plt.grid(True)
```
