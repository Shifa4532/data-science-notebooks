```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
print(tips.head())
```

```python
sns.set_theme(style="darkgrid")
sns.relplot(
    data =tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python
dia=sns.load_dataset("diamonds")
print(dia.head())
```

```python
sns.scatterplot(
    data=dia,
    x="price",
    y="carat",
    hue="cut"
)
```

```python
fly=sns.load_dataset("flights")
print(fly.head())
```

```python
sns.lineplot(
    data=fly,
    x="year",
    y="passengers"
)
```

```python
sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
sns.boxplot(
    data=tips,
    x="day",
    y="tip",
)
```

```python
pengy = sns.load_dataset("penguins")
pengy.head()
```

```python
sns.histplot(
    data=pengy,
    x="body_mass_g",
    bins=30,
    color="pink"
)
```

```python
pivoted_data=fly.pivot(index="month",columns="year",values="passengers")
```

```python
sns.heatmap(
    pivoted_data,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)
plt.title("passengers heatmap")
plt.tight_layout()
```

```python
fig, ax = plt.subplots()
sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    ax=ax,
    marker=".",
    errorbar=None
)

ax.set_ylabel("total bill amt")
ax.set_xlabel("dayz")
ax.set_title("avg bill value for each day")
```
